```python
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt
from IPython.display import display
import numpy as np
from qiskit import QuantumCircuit
```

```python
# READING BLOTCH SPHERE
# creates a row of individual Bloch spheres—one for each qubit
# Z axis (classical states)
# -> UP / +Z = exactly |0>
# -> DOWN / -Z = exactly |1>
# Equator (superposition states)
# -> FRONT / +X = |+> = (|0> + |1>)/2
# -> BACK / -X = |-> = (|0> - |1>)/2
# -> RIGHT / +Y = |i+> = (|0> + i|1>)/2
# -> LEFT / -Y = |i-> = (|0> - i|1>)/2
# Arrow touching the surface : length=1, pure state
# Dot : entanglement with another qubit

qc = QuantumCircuit(2)
qc.h(0)
qc.x(1)

display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))
```

```python
# CONTROLLED-X GATE (bit flip/pauli x gate when control qubit is |1>)

qc = QuantumCircuit(2)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

qc.h(0)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

qc.cx(0, 1)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

display(qc.draw('mpl'))
```

```python
# CONTROLLED-Y GATE (bit flip + phase flip/pauli y gate when control qubit is |1>)

qc = QuantumCircuit(2)
display(Statevector.from_instruction(qc).draw('latex'))

qc.x(0)       # Control = 1
display(Statevector.from_instruction(qc).draw('latex'))

qc.cy(0, 1)   # Controlled-Y
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

display(qc.draw('mpl'))
```

```python
# CONTROLLED-Z GATE (phase flip/pauli z gate when control qubit is |1>)

qc = QuantumCircuit(2)
display(Statevector.from_instruction(qc).draw('latex'))

qc.h(0)
display(Statevector.from_instruction(qc).draw('latex'))

qc.h(1)
display(Statevector.from_instruction(qc).draw('latex'))

qc.cz(0, 1) # changes phase of |11> --> -|11>
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

display(qc.draw('mpl'))
```

```python
# CONTROLLED-H GATE

qc = QuantumCircuit(2)
display(Statevector.from_instruction(qc).draw('latex'))

qc.x(0)
display(Statevector.from_instruction(qc).draw('latex'))

qc.ch(0, 1)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

display(qc.draw('mpl'))
```

```python
# CONTROLLED-UNITARY GATE

qc = QuantumCircuit(2)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

theta = np.pi / 2
phi = 0
lam = np.pi
gamma = 0

qc.h(0)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

qc.cu(theta, phi, lam, gamma, 0, 1)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

display(qc.draw('mpl'))
```

```python
# TOFOLLI / CCX GATE (target flips only when control qubits are |11>)

qc = QuantumCircuit(3)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

qc.x(0)
qc.x(1)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

qc.ccx(0, 1, 2)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

display(qc.draw('mpl'))
```

```python
# CONTROLLED-SWAP (FREDKIN) GATE (swaps two target qubits when control qubit is 1)

qc = QuantumCircuit(3)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

qc.x(0)   # Control = 1
qc.x(2)   # Target 2 = 1
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

qc.cswap(0, 1, 2)
display(Statevector.from_instruction(qc).draw('latex'))
display(plot_bloch_multivector(Statevector.from_instruction(qc)))

display(qc.draw('mpl'))
```
