```python
import numpy as np
from qiskit import QuantumCircuit, QuantumRegister
from qiskit.quantum_info import Statevector
from IPython.display import display
```

```python
value = QuantumRegister(1, "value")
feature = QuantumRegister(2, "feature")
row = QuantumRegister(4, "row")

qc = QuantumCircuit(value, feature, row)

# qc.x(value[0])

qc.x(feature[1])

qc.x(row[1])
qc.x(row[3])

display(qc.draw("mpl"))
```

```python
state = Statevector.from_instruction(qc)

print("\nStatevector:")
display(state.draw("latex"))
```

Qiskit is called little-endian because it treats qubit q0 as the least significant bit (LSB)

## **Little-Endian (LSB first)**

State = |q6 q5 q4 q3 q2 q1 q0> = |row3> | row2> |row1> |row0> |feature1> |feature0> |value>

State = |r3 r2 r1 r0 f1 f0 v>

Feature = 01   [f0 f1]

Row = 0101     [r0 r1 r2 r3]

## **Big-Endian (MSB first)**

State = |q6 q5 q4 q3 q2 q1 q0> = |row3> | row2> |row1> |row0> |feature1> |feature0> |value>

State = |r3 r2 r1 r0 f1 f0 v>

Feature = 10    [f1 f0]

Row = 1010      [r3 r2 r1 r0]
