x0 and x1
0 and 0 = 0
0 and 1 = 0
1 and 0 = 0
1 and 1 = 1

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
```

```python
def boolean_and(x1, x0):
    qc = QuantumCircuit(3, 1)

    if x1:
        qc.x(0)

    if x0:
        qc.x(1)

    # Toffoli gate - if both are 1 then flip the output qubit to 1
    qc.ccx(0, 1, 2)

    # Measure output
    qc.measure(2, 0)

    simulator = AerSimulator()
    result = simulator.run(transpile(qc, simulator), shots=1).result()
    ans = next(iter(result.get_counts()))
    display(qc.draw(output='mpl'))

    print(f"{x1} and {x0} = {ans}")
```

```python
boolean_and(0, 0)
```

```python
boolean_and(0, 1)
```

```python
boolean_and(1, 0)
```

```python
boolean_and(1, 1)
```
