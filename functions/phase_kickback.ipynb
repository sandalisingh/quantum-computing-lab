Uf |x> |-> = (-1)^f(x) |x> |->

For AND:
- Input 00 gets phase +1
- Input 01 gets phase +1
- Input 10 gets phase +1
- Input 11 gets phase -1

This is called phase kickback.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
qc = QuantumCircuit(3, 2)

# Superposition of input values
qc.h(0)
qc.h(1)
display(Statevector(qc).draw('latex'))
```

```python
# Prepare output in |-> state
qc.x(2)
qc.h(2)
display(Statevector(qc).draw('latex'))
```

```python
# AND oracle
qc.ccx(0, 1, 2)
display(Statevector(qc).draw('latex'))
```

```python
# Measure input qubits only
qc.measure([0, 1], [0, 1])
```

```python
display(qc.draw(output='mpl'))
```
